**Cryptocurrency Market Analysis: Historical Behavior, Risk Assessment, and Price Forecasting for BTC, SOL, and TRX (2010–2030)**

This project analyzes three cryptocurrencies — Bitcoin (BTC), Solana (SOL),
and TRON (TRX) — using historical daily price data from Kaggle. The dataset
covers different time periods for each coin: BTC from 2010, TRX from 2017,
and SOL from 2020, all up to March 2026.

The analysis is divided into several parts. First, we look at the full price
history of each coin and identify patterns by month, week, and season. Then
we study how the three coins relate to each other using correlation analysis.
We also test a popular hypothesis in crypto: that TRX behaves differently
from other coins when Bitcoin crashes. After that, we build buy signal
indicators based on historical data and run Monte Carlo simulations to
estimate where prices could be by 2027, 2028, 2029, and 2030.

The goal is not to give financial advice, but to use data to understand
the behavior of these coins and give a more informed view of what could
happen in the future.

sol : https://www.kaggle.com/datasets/svaningelgem/crypto-currencies-daily-prices?select=SOL.csv

btc: https://www.kaggle.com/datasets/svaningelgem/crypto-currencies-daily-prices?select=BTC.csv

trx: https://www.kaggle.com/datasets/svaningelgem/crypto-currencies-daily-prices?select=TRX.csv


## Results Analysis

### Price History

Bitcoin started at less than one dollar in 2010 and reached an all-time high of **`$125,173`**. At the time of this analysis its price was around **`$66,000`**. Solana started trading in April 2020 at under `$1` and reached a high of **`$258`**. TRON has been trading since 2017 and its all-time high is only **`$0.37`**, which is very close to its current price of `$0.31`. This already tells us something important: TRX moves in a very different price range and has a completely different growth story compared to BTC and SOL.

### Monthly Variation

The best month to buy BTC historically has been **February**, while **September** is the worst. This is actually a well known pattern in crypto markets. September tends to be bearish for Bitcoin almost every year. For SOL, **August** has been the best entry point and **May** the worst. TRX works differently: **December** is the best month to buy it, probably connected to year-end market movements, and **June** is consistently the worst.

### Trend Duration

The analysis looked at how many days it takes each coin to move 5% in either direction from any starting point. All three coins move fast in both directions, which reflects the high volatility of the crypto market in general. The distributions show that most significant moves happen within a few days, and there is no big difference between the time it takes to gain or lose value.

### Correlation Analysis

This is one of the most interesting parts of the analysis. The correlation between daily returns was calculated using only the dates where all three coins had data at the same time. The results were:

| Pair | Correlation |
|------|------------|
| BTC and SOL | 0.5218 (moderate positive) |
| BTC and TRX | 0.4933 (moderate positive) |
| SOL and TRX | 0.3500 (weak positive) |

These numbers tell us that all three coins move in the same general direction when Bitcoin moves, but the relationship is not strong enough to say they are identical. SOL follows BTC a bit more closely than TRX does.

### TRX Anomaly Test

One of the main questions of this project was whether TRX resists Bitcoin crashes better than other coins. The data gives a clear answer: **no, it does not**. When BTC fell more than 3% in a single day (230 events in the dataset), TRX was only positive **11.7%** of the time, and its average return on those days was **-4.03%**. SOL was positive **14.8%** of the time with an average of **-5.65%** on the same days. So TRX actually falls with BTC, just slightly less than SOL in percentage terms on average. The rolling 90 day correlation chart shows that BTC and TRX have periods of lower correlation, which might be the source of that perception, but it does not mean TRX is immune to Bitcoin crashes.

### Buy Signal Analysis

Three signals were tested for each coin: best historical month, what happens 30 days after a single day drop of 8% or more, and what happens 30 days after the RSI drops below 30.

After an 8% drop in one day, the results are actually encouraging:

| Coin | Avg 30 day return | Profitable events |
|------|------------------|------------------|
| BTC | +27.2% | 58% |
| SOL | +23.0% | 50% |
| TRX | +38.6% | 51% |

TRX showed the highest average recovery, although only 51% of events were profitable. This means the average is pulled up by a few very large recoveries rather than consistent gains.

The RSI signal was weaker overall. When RSI dropped below 30, BTC recovered **6.3%** on average but only **49%** of events were positive. TRX had the lowest average at **2.8%** but the highest success rate at **60%**, meaning the gains were small but more consistent.

### Monte Carlo Forecast 2027 to 2030

The simulation used 1,000 scenarios for each coin based on historical daily log returns. The results show a very wide range of outcomes, which is normal for assets this volatile.

**Bitcoin (BTC) — current price `$66,000`**

| Year | Bear scenario | Base scenario | Bull scenario |
|------|--------------|--------------|--------------|
| 2027 | `$34,338` | `$338,283` | `$2,992,280` |
| 2028 | `$46,911` | `$859,242` | `$13,852,993` |
| 2029 | `$84,989` | `$2,205,525` | `$57,099,273` |

**Solana (SOL) — current price `$82.72`**

| Year | Bear scenario | Base scenario | Bull scenario |
|------|--------------|--------------|--------------|
| 2027 | `$25.76` | `$292.74` | `$3,475.39` |
| 2028 | `$22.65` | `$612.52` | `$15,747.39` |
| 2029 | `$26.53` | `$1,539.04` | `$50,092.65` |

**TRON (TRX) — current price `$0.31`**

| Year | Bear scenario | Base scenario | Bull scenario |
|------|--------------|--------------|--------------|
| 2027 | `$0.07` | `$0.87` | `$10.95` |
| 2028 | `$0.08` | `$1.42` | `$33.19` |
| 2029 | `$0.08` | `$2.66` | `$91.00` |

These numbers look extreme in the bull case but they reflect what the historical volatility of each coin actually implies when projected forward over several years.

## Conclusions

### Which coin has the highest potential appreciation?

In terms of percentage gain relative to current price, TRX has the most extreme upside in the bull scenario. Going from `$0.31` to `$91` by 2029 represents a gain of almost **29,000%**. SOL going from `$82` to `$50,092` is also a massive move. BTC going from `$66,000` to over `$57` million sounds extraordinary, but that is what 15 years of historical volatility implies when you run 1,000 random scenarios forward.

However, the **base scenario** is a better reference point than the bull case. In the base scenario, SOL shows the most attractive percentage gain relative to its current price, going from `$82` to around `$1,539` by 2029. That is roughly **18x**. BTC in the base case goes from `$66,000` to about `$2.2` million, which is **33x** but requires a much larger initial investment. TRX in the base case goes from `$0.31` to `$2.66`, roughly **8x**.

### Which coin would make sense to buy and where?

**SOL** is probably the most balanced option if the goal is maximum potential with a manageable amount of capital. It has a real ecosystem with active developers and projects built on it, moderate correlation with BTC (0.52), and a base case forecast that implies strong growth. SOL can be purchased on Binance, Coinbase, Kraken, or directly through wallets like Phantom.

**BTC** is the safest option in relative terms. It has the longest history, the most liquidity, and the strongest institutional backing. If the available capital is limited, buying small amounts regularly during months like February or after large drops has historically been a solid strategy. BTC is available on virtually every exchange worldwide including Binance, Coinbase, Kraken and Bitso.

**TRX** is interesting as a small speculative position. Its correlation with BTC is the lowest of the three (0.49), which means it partially decorrelates from the rest of the portfolio. Its price is so low that even a small investment gives you many tokens, and in the bull scenario the upside is extraordinary. However, the base case recovery after negative signals is weak compared to the other two coins. TRON is available on Binance, Huobi and OKX, and it has very low transaction fees which makes it practical for people who want to move funds quickly on the blockchain.

### What about correlations?

The moderate correlation between all three coins (between 0.35 and 0.52) means they move in the same direction most of the time, especially when BTC has a strong move. This is important to understand: **holding all three does not give true diversification within crypto**. When Bitcoin has a bad week, all three will likely go down together. True diversification would require including assets outside of the crypto market.

That said, the correlation is not perfect. There are periods visible in the rolling correlation chart where TRX decorrelates from BTC. These windows could be interesting for short term traders but are not predictable enough to build a reliable strategy around them.

### On the TRX hypothesis

The idea that TRX resists Bitcoin crashes is not supported by the data. When BTC falls more than 3% in a day, TRX is positive only 11.7% of the time and loses an average of 4% on those same days. This is actually less negative than SOL on average, but the difference is not large enough to call TRX an anomaly. What the data does suggest is that TRX has **lower sensitivity** to BTC moves in general (correlation of 0.49 vs 0.52 for SOL), but it still follows Bitcoin in the same direction the vast majority of the time.

### Final thoughts

The bear scenarios in the Monte Carlo simulation are real possibilities. BTC could be at `$34,000` in 2027. SOL could fall back to `$25`. TRX could drop below `$0.10`. Crypto is still a high risk asset class and none of these forecasts should be taken as certainty. The models are based only on historical price data and do not account for regulation, technology changes, competition between blockchains, or macroeconomic conditions.

If the goal is long term holding until 2030, the historical data suggests that buying during moments of fear, large drops, oversold RSI readings, or historically weak months like September for BTC or June for TRX , has produced better outcomes than buying at random times. That pattern is consistent across all three coins and it is probably the most actionable takeaway from this entire analysis.



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from scipy import stats

plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor']   = '#0d1117'
plt.rcParams['axes.edgecolor']   = '#30363d'
plt.rcParams['axes.labelcolor']  = 'white'
plt.rcParams['xtick.color']      = 'white'
plt.rcParams['ytick.color']      = 'white'
plt.rcParams['text.color']       = 'white'
plt.rcParams['axes.grid']        = True
plt.rcParams['grid.alpha']       = 0.2
plt.rcParams['grid.color']       = '#30363d'

COLORS = {'BTC': '#F7931A', 'SOL': '#9945FF', 'TRX': '#FF0013'}
COINS  = ['BTC', 'SOL', 'TRX']
MONTH_NAMES = ['Jan','Feb','Mar','Apr','May','Jun',
               'Jul','Aug','Sep','Oct','Nov','Dec']

print("Libraries loaded.")


btc = pd.read_csv('/content/BTC.csv')
sol = pd.read_csv('/content/SOL.csv')
trx = pd.read_csv('/content/TRX.csv')

raw = {'BTC': btc, 'SOL': sol, 'TRX': trx}
data = {}

for coin, df in raw.items():
    df               = df.copy()
    df['date']       = pd.to_datetime(df['date'])
    df               = df.sort_values('date').reset_index(drop=True)
    df['return']     = df['close'].pct_change()
    df['return_pct'] = df['return'] * 100
    df['log_return'] = np.log(df['close'] / df['close'].shift(1))
    df['rolling_30'] = df['close'].rolling(30).mean()
    df['rolling_200']= df['close'].rolling(200).mean()
    df['year']       = df['date'].dt.year
    df['month']      = df['date'].dt.month
    df['week']       = df['date'].dt.isocalendar().week.astype(int)
    data[coin]       = df

# Aligned returns on common dates
btc_r = data['BTC'].set_index('date')['return'].dropna()
sol_r = data['SOL'].set_index('date')['return'].dropna()
trx_r = data['TRX'].set_index('date')['return'].dropna()

common_idx  = btc_r.index.intersection(sol_r.index).intersection(trx_r.index)
returns_df  = pd.DataFrame({
    'BTC': btc_r[common_idx],
    'SOL': sol_r[common_idx],
    'TRX': trx_r[common_idx]
}).dropna()

corr_matrix = returns_df.corr()

print("Data loaded:")
for coin in COINS:
    df = data[coin]
    print(f"  {coin}: {df['date'].min().date()} to {df['date'].max().date()} "
          f"| current: ${df['close'].iloc[-1]:,.4f}")


fig, axes = plt.subplots(3, 1, figsize=(18, 16))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('Historical Price  –  BTC | SOL | TRX',
             fontsize=18, fontweight='bold', color='white', y=0.98)

for i, coin in enumerate(COINS):
    df  = data[coin]
    ax  = axes[i]
    col = COLORS[coin]
    ax.set_facecolor('#0d1117')

    ax.fill_between(df['date'], df['close'], alpha=0.15, color=col)
    ax.plot(df['date'], df['close'],      color=col,     linewidth=1.2, label='Price')
    ax.plot(df['date'], df['rolling_30'], color='white', linewidth=1,
            linestyle='--', alpha=0.6, label='MA 30')
    ax.plot(df['date'], df['rolling_200'],color='yellow',linewidth=1,
            linestyle='--', alpha=0.6, label='MA 200')

    ath_idx = df['close'].idxmax()
    ax.annotate(f"ATH ${df['close'].max():,.2f}",
                xy=(df['date'].iloc[ath_idx], df['close'].max()),
                xytext=(df['date'].iloc[ath_idx], df['close'].max() * 0.72),
                arrowprops=dict(arrowstyle='->', color='white', lw=1.2),
                fontsize=8, color='white',
                bbox=dict(boxstyle='round,pad=0.3', facecolor='#1f2937', alpha=0.8))

    ax.set_title(f"{coin}  –  Current: ${df['close'].iloc[-1]:,.4f}",
                 fontsize=13, fontweight='bold', color=col, pad=8)
    ax.set_ylabel('Price (USD)', fontsize=10)
    ax.legend(fontsize=9, loc='upper left')

axes[-1].set_xlabel('Date', fontsize=11)
plt.tight_layout()
plt.savefig('01_price_history.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print("Figure 1: Price History saved.")


fig, axes = plt.subplots(1, 3, figsize=(20, 7))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('Average Monthly Price Variation (%)',
             fontsize=16, fontweight='bold', color='white')

for i, coin in enumerate(COINS):
    df  = data[coin]
    ax  = axes[i]
    col = COLORS[coin]
    ax.set_facecolor('#0d1117')

    monthly = df.groupby('month')['return_pct'].mean().reindex(range(1, 13))
    bar_cols = [col if v >= 0 else '#ef4444' for v in monthly.values]
    bars = ax.bar(range(1, 13), monthly.values, color=bar_cols,
                  edgecolor='#0d1117', linewidth=0.8)

    for bar, val in zip(bars, monthly.values):
        if not np.isnan(val):
            ax.text(bar.get_x() + bar.get_width() / 2,
                    bar.get_height() + (0.04 if val >= 0 else -0.15),
                    f'{val:.2f}%', ha='center', fontsize=7.5,
                    color='white', fontweight='bold')

    ax.axhline(0, color='white', linewidth=0.8, alpha=0.5)
    ax.set_xticks(range(1, 13))
    ax.set_xticklabels(MONTH_NAMES, fontsize=9, rotation=45)
    ax.set_title(coin, fontsize=14, fontweight='bold', color=col)
    ax.set_ylabel('Avg Daily Return (%)', fontsize=9)

    best  = monthly.idxmax()
    worst = monthly.idxmin()
    ax.text(0.98, 0.97,
            f"Best:  {MONTH_NAMES[best-1]} ({monthly[best]:+.2f}%)\n"
            f"Worst: {MONTH_NAMES[worst-1]} ({monthly[worst]:+.2f}%)",
            transform=ax.transAxes, fontsize=8, color='white',
            va='top', ha='right',
            bbox=dict(boxstyle='round', facecolor='#1f2937', alpha=0.8))

plt.tight_layout()
plt.savefig('02_monthly_variation.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print("Figure 2: Monthly Variation saved.")



def trend_streaks(df, threshold=0.05):
    close   = df['close'].values
    up, dn  = [], []
    i = 0
    while i < len(close) - 1:
        start = close[i]
        j = i + 1
        while j < len(close):
            chg = (close[j] - start) / start
            if abs(chg) >= threshold:
                (up if chg > 0 else dn).append(j - i)
                i = j
                break
            j += 1
        else:
            break
    return np.array(up), np.array(dn)

fig, axes = plt.subplots(2, 3, figsize=(20, 10))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('Days needed to move +5% or -5% from any starting point',
             fontsize=14, fontweight='bold', color='white')

for ci, coin in enumerate(COINS):
    df    = data[coin]
    color = COLORS[coin]
    up, dn = trend_streaks(df, threshold=0.05)

    ax_top = axes[0, ci]
    ax_top.set_facecolor('#0d1117')
    if len(up) > 0:
        ax_top.hist(up, bins=30, alpha=0.75, color=color,
                    label=f'Up  avg {up.mean():.1f}d', edgecolor='#0d1117')
    if len(dn) > 0:
        ax_top.hist(dn, bins=30, alpha=0.55, color='#ef4444',
                    label=f'Down avg {dn.mean():.1f}d', edgecolor='#0d1117')
    ax_top.set_title(f'{coin} – Trend duration', fontsize=12,
                     fontweight='bold', color=color)
    ax_top.set_xlabel('Days', fontsize=9)
    ax_top.set_ylabel('Frequency', fontsize=9)
    ax_top.legend(fontsize=9)

    ax_bot = axes[1, ci]
    ax_bot.set_facecolor('#0d1117')
    avgs = [up.mean() if len(up) > 0 else 0, dn.mean() if len(dn) > 0 else 0]
    meds = [np.median(up) if len(up) > 0 else 0, np.median(dn) if len(dn) > 0 else 0]
    x    = np.arange(2)
    ax_bot.bar(x - 0.2, avgs, 0.35, label='Average', color=color,   alpha=0.85)
    ax_bot.bar(x + 0.2, meds, 0.35, label='Median',  color='white', alpha=0.4)
    ax_bot.set_xticks(x)
    ax_bot.set_xticklabels(['Up trends', 'Down trends'], fontsize=10)
    ax_bot.set_ylabel('Days', fontsize=9)
    ax_bot.set_title(f'{coin} – Avg vs Median', fontsize=11, color=color)
    ax_bot.legend(fontsize=9)
    for xi, (a, m) in enumerate(zip(avgs, meds)):
        ax_bot.text(xi - 0.2, a + 0.1, f'{a:.1f}d', ha='center',
                    fontsize=9, color='white', fontweight='bold')
        ax_bot.text(xi + 0.2, m + 0.1, f'{m:.1f}d', ha='center',
                    fontsize=9, color='white', fontweight='bold')

plt.tight_layout()
plt.savefig('03_trend_duration.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print("Figure 3: Trend Duration saved.")


fig, axes = plt.subplots(1, 3, figsize=(20, 6))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('Correlation Analysis – BTC | SOL | TRX',
             fontsize=16, fontweight='bold', color='white')

ax = axes[0]
ax.set_facecolor('#0d1117')
sns.heatmap(corr_matrix, annot=True, fmt='.3f', cmap='RdYlGn',
            ax=ax, linewidths=1, linecolor='#0d1117',
            vmin=-1, vmax=1, square=True,
            annot_kws={'size': 14, 'weight': 'bold'})
ax.set_title('Pearson Correlation\n(daily returns – common dates)',
             fontsize=12, fontweight='bold', color='white')
ax.tick_params(colors='white')

for ax_i, coin_b, ax_obj in [('SOL', 'SOL', axes[1]), ('TRX', 'TRX', axes[2])]:
    ax_obj.set_facecolor('#0d1117')
    x_vals = returns_df['BTC'].values
    y_vals = returns_df[coin_b].values
    ax_obj.scatter(x_vals, y_vals, alpha=0.3,
                   color=COLORS[coin_b], s=8, edgecolors='none')
    z  = np.polyfit(x_vals, y_vals, 1)
    p  = np.poly1d(z)
    xl = np.linspace(x_vals.min(), x_vals.max(), 100)
    ax_obj.plot(xl, p(xl), color='white', linewidth=2, linestyle='--')
    r  = corr_matrix.loc['BTC', coin_b]
    ax_obj.set_title(f'BTC vs {coin_b}  (r = {r:.3f})',
                     fontsize=12, fontweight='bold', color='white')
    ax_obj.set_xlabel('BTC daily return', fontsize=10)
    ax_obj.set_ylabel(f'{coin_b} daily return', fontsize=10)
    ax_obj.axhline(0, color='white', alpha=0.3, linewidth=0.8)
    ax_obj.axvline(0, color='white', alpha=0.3, linewidth=0.8)

plt.tight_layout()
plt.savefig('04_correlation.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()

print("\nCorrelation matrix:")
print(corr_matrix.round(4))
print("Figure 4: Correlation saved.")


btc_crash_mask = returns_df['BTC'] < -0.03
btc_crash_days = returns_df[btc_crash_mask].index

trx_on_crash   = returns_df.loc[btc_crash_days, 'TRX']
sol_on_crash   = returns_df.loc[btc_crash_days, 'SOL']
pct_trx_pos    = (trx_on_crash > 0).mean() * 100
pct_sol_pos    = (sol_on_crash > 0).mean() * 100

rolling_corr_trx = returns_df['BTC'].rolling(90).corr(returns_df['TRX'])
rolling_corr_sol = returns_df['BTC'].rolling(90).corr(returns_df['SOL'])

fig, axes = plt.subplots(1, 2, figsize=(18, 7))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('TRX Anomaly – Does TRX resist BTC crashes?',
             fontsize=15, fontweight='bold', color='white')

ax1 = axes[0]
ax1.set_facecolor('#0d1117')
ax1.hist(trx_on_crash * 100, bins=30, color=COLORS['TRX'],
         alpha=0.8, edgecolor='#0d1117', label='TRX')
ax1.hist(sol_on_crash * 100, bins=30, color=COLORS['SOL'],
         alpha=0.5, edgecolor='#0d1117', label='SOL')
ax1.axvline(0, color='white', linewidth=1.5, linestyle='--')
ax1.axvline(trx_on_crash.mean() * 100, color=COLORS['TRX'], linewidth=2,
            label=f'TRX avg {trx_on_crash.mean()*100:+.2f}%')
ax1.axvline(sol_on_crash.mean() * 100, color=COLORS['SOL'], linewidth=2,
            linestyle=':', label=f'SOL avg {sol_on_crash.mean()*100:+.2f}%')
ax1.set_title('TRX vs SOL return when BTC falls > 3%',
              fontsize=11, fontweight='bold', color='white')
ax1.set_xlabel('Daily return (%)', fontsize=10)
ax1.set_ylabel('Frequency', fontsize=10)
ax1.legend(fontsize=9)
ax1.text(0.02, 0.97,
         f"BTC crash days (< -3%): {len(btc_crash_days)}\n"
         f"TRX positive: {(trx_on_crash>0).sum()} ({pct_trx_pos:.1f}%)\n"
         f"SOL positive: {(sol_on_crash>0).sum()} ({pct_sol_pos:.1f}%)",
         transform=ax1.transAxes, fontsize=9, color='white', va='top',
         bbox=dict(boxstyle='round', facecolor='#1f2937', alpha=0.8))

ax2 = axes[1]
ax2.set_facecolor('#0d1117')
ax2.fill_between(rolling_corr_trx.index, rolling_corr_trx.values,
                 where=(rolling_corr_trx.values >= 0),
                 alpha=0.4, color=COLORS['BTC'])
ax2.fill_between(rolling_corr_trx.index, rolling_corr_trx.values,
                 where=(rolling_corr_trx.values < 0),
                 alpha=0.4, color=COLORS['TRX'])
ax2.plot(rolling_corr_trx.index, rolling_corr_trx.values,
         color=COLORS['TRX'], linewidth=1.2, label='BTC-TRX')
ax2.plot(rolling_corr_sol.index, rolling_corr_sol.values,
         color=COLORS['SOL'], linewidth=1.2, alpha=0.7, label='BTC-SOL')
ax2.axhline(0, color='white', linewidth=0.8, linestyle='--', alpha=0.5)
ax2.set_title('Rolling 90-day Correlation: BTC vs TRX and SOL',
              fontsize=11, fontweight='bold', color='white')
ax2.set_xlabel('Date', fontsize=10)
ax2.set_ylabel('Correlation', fontsize=10)
ax2.set_ylim(-1, 1)
ax2.legend(fontsize=10)

plt.tight_layout()
plt.savefig('05_trx_anomaly.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()

print(f"\nTRX anomaly result:")
print(f"  BTC crash days (< -3%): {len(btc_crash_days)}")
print(f"  TRX positive on those days: {pct_trx_pos:.1f}%")
print(f"  SOL positive on those days: {pct_sol_pos:.1f}%")
print(f"  TRX avg return on crash days: {trx_on_crash.mean()*100:+.2f}%")
print(f"  SOL avg return on crash days: {sol_on_crash.mean()*100:+.2f}%")
print("Figure 5: TRX Anomaly saved.")


fig, axes = plt.subplots(2, 2, figsize=(18, 12))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('Volatility and Risk Analysis',
             fontsize=16, fontweight='bold', color='white')

ax1 = axes[0, 0]
ax1.set_facecolor('#0d1117')
for coin in COINS:
    df  = data[coin]
    vol = df['return'].rolling(90).std() * np.sqrt(365) * 100
    ax1.plot(df['date'], vol, color=COLORS[coin], linewidth=1.2,
             label=coin, alpha=0.85)
ax1.set_title('Rolling 90-day Annualized Volatility (%)',
              fontsize=12, fontweight='bold', color='white')
ax1.set_ylabel('Volatility (%)')
ax1.legend(fontsize=11)

ax2 = axes[0, 1]
ax2.set_facecolor('#0d1117')
for coin in COINS:
    df = data[coin].dropna(subset=['return_pct'])
    ax2.hist(df['return_pct'], bins=80, alpha=0.45, color=COLORS[coin],
             label=coin, density=True, edgecolor='none')
    ax2.axvline(df['return_pct'].mean(), color=COLORS[coin],
                linewidth=1.5, linestyle='--')
ax2.set_xlim(-30, 30)
ax2.set_title('Daily Return Distribution',
              fontsize=12, fontweight='bold', color='white')
ax2.set_xlabel('Daily Return (%)')
ax2.set_ylabel('Density')
ax2.legend(fontsize=11)

ax3 = axes[1, 0]
ax3.set_facecolor('#0d1117')
for coin in COINS:
    df       = data[coin]
    roll_max = df['close'].cummax()
    drawdown = (df['close'] - roll_max) / roll_max * 100
    ax3.fill_between(df['date'], drawdown, 0, alpha=0.25, color=COLORS[coin])
    ax3.plot(df['date'], drawdown, color=COLORS[coin], linewidth=0.8, label=coin)
ax3.set_title('Drawdown from All-Time High (%)',
              fontsize=12, fontweight='bold', color='white')
ax3.set_xlabel('Date')
ax3.set_ylabel('Drawdown (%)')
ax3.legend(fontsize=11)

ax4 = axes[1, 1]
ax4.set_facecolor('#0d1117')
ax4.axis('off')
table_data = []
for coin in COINS:
    df       = data[coin].dropna(subset=['return'])
    roll_max = df['close'].cummax()
    dd       = (df['close'] - roll_max) / roll_max * 100
    table_data.append([
        coin,
        f"{df['return_pct'].mean():.3f}%",
        f"{df['return_pct'].std():.2f}%",
        f"{df['return_pct'].max():.2f}%",
        f"{df['return_pct'].min():.2f}%",
        f"{dd.min():.1f}%"
    ])
col_labels = ['Coin','Avg Return','Std Dev','Best Day','Worst Day','Max DD']
table = ax4.table(cellText=table_data, colLabels=col_labels,
                  loc='center', cellLoc='center')
table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1.2, 2.2)
for (row, col), cell in table.get_celld().items():
    cell.set_facecolor('#1f2937' if row == 0 else '#0d1117')
    cell.set_edgecolor('#30363d')
    cell.set_text_props(color='white',
                        fontweight='bold' if row == 0 else 'normal')
ax4.set_title('Risk Summary', fontsize=12, fontweight='bold',
              color='white', pad=15)

plt.tight_layout()
plt.savefig('06_volatility_risk.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print("Figure 6: Volatility and Risk saved.")


fig, axes = plt.subplots(1, 3, figsize=(20, 7))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('Annual Performance by Coin (%)',
             fontsize=16, fontweight='bold', color='white')

for i, coin in enumerate(COINS):
    df    = data[coin]
    color = COLORS[coin]
    ax    = axes[i]
    ax.set_facecolor('#0d1117')

    yearly = df.groupby('year').apply(
        lambda x: (x['close'].iloc[-1] / x['close'].iloc[0] - 1) * 100
    ).reset_index()
    yearly.columns = ['year', 'return_pct']

    bar_cols = [color if v >= 0 else '#ef4444' for v in yearly['return_pct']]
    bars = ax.bar(yearly['year'].astype(str), yearly['return_pct'],
                  color=bar_cols, edgecolor='#0d1117', linewidth=0.8)

    for bar, val in zip(bars, yearly['return_pct']):
        offset = max(abs(val) * 0.03, 200)
        ax.text(bar.get_x() + bar.get_width() / 2,
                bar.get_height() + (offset if val >= 0 else -offset * 6),
                f'{val:+,.0f}%', ha='center', fontsize=7.5,
                color='white', fontweight='bold', rotation=45)

    ax.axhline(0, color='white', linewidth=0.8, alpha=0.5)
    ax.set_title(coin, fontsize=14, fontweight='bold', color=color)
    ax.set_xlabel('Year', fontsize=10)
    ax.set_ylabel('Annual Return (%)', fontsize=10)
    ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('07_yearly_performance.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print("Figure 7: Yearly Performance saved.")



def buy_signal_analysis(df, coin, color, drop_threshold=-0.08, fwd_days=30):
    results = {}

    # After a one-day drop of X%, what happens next N days?
    drops = df[df['return'] <= drop_threshold].index
    future_rets = []
    for idx in drops:
        loc = df.index.get_loc(idx)
        if loc + fwd_days < len(df):
            fwd = (df['close'].iloc[loc + fwd_days] / df['close'].iloc[loc] - 1) * 100
            future_rets.append(fwd)
    results['drop_signal'] = future_rets

    # Best months historically
    monthly_avg = df.groupby('month')['return_pct'].mean()
    results['best_month']  = monthly_avg.idxmax()
    results['worst_month'] = monthly_avg.idxmin()
    results['monthly_avg'] = monthly_avg

    # Best weeks historically
    weekly_avg = df.groupby('week')['return_pct'].mean()
    results['best_weeks']  = weekly_avg.nlargest(5).index.tolist()
    results['worst_weeks'] = weekly_avg.nsmallest(5).index.tolist()

    # RSI (simplified: 14-day)
    delta     = df['close'].diff()
    gain      = delta.clip(lower=0).rolling(14).mean()
    loss      = (-delta.clip(upper=0)).rolling(14).mean()
    rs        = gain / loss
    df['rsi'] = 100 - (100 / (1 + rs))
    oversold_days = df[df['rsi'] < 30]
    over_rets = []
    for idx in oversold_days.index:
        loc = df.index.get_loc(idx)
        if loc + fwd_days < len(df):
            fwd = (df['close'].iloc[loc + fwd_days] / df['close'].iloc[loc] - 1) * 100
            over_rets.append(fwd)
    results['rsi_signal'] = over_rets

    return results

fig, axes = plt.subplots(3, 3, figsize=(22, 18))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('Buy Signal Analysis – When is the best time to buy?',
             fontsize=16, fontweight='bold', color='white', y=0.98)

signal_summary = {}

for ci, coin in enumerate(COINS):
    df    = data[coin].copy().reset_index(drop=True)
    color = COLORS[coin]
    res   = buy_signal_analysis(df, coin, color)
    signal_summary[coin] = res

    # Row 0: Best month to buy (monthly avg return heatmap)
    ax0 = axes[0, ci]
    ax0.set_facecolor('#0d1117')
    monthly = res['monthly_avg'].reindex(range(1, 13))
    bar_c   = [color if v >= 0 else '#ef4444' for v in monthly.values]
    ax0.bar(range(1, 13), monthly.values, color=bar_c, edgecolor='#0d1117')
    ax0.axhline(0, color='white', linewidth=0.8, alpha=0.5)
    ax0.set_xticks(range(1, 13))
    ax0.set_xticklabels(MONTH_NAMES, fontsize=8, rotation=45)
    ax0.set_title(f'{coin} – Avg return by month\n'
                  f'Best: {MONTH_NAMES[res["best_month"]-1]}  '
                  f'Worst: {MONTH_NAMES[res["worst_month"]-1]}',
                  fontsize=11, fontweight='bold', color=color)
    ax0.set_ylabel('Avg daily return (%)', fontsize=9)

    # Row 1: After -8% drop, what happens in next 30 days
    ax1 = axes[1, ci]
    ax1.set_facecolor('#0d1117')
    drop_arr = np.array(res['drop_signal'])
    if len(drop_arr) > 0:
        ax1.hist(drop_arr, bins=25, color=color, alpha=0.8, edgecolor='#0d1117')
        ax1.axvline(0, color='white', linewidth=1.5, linestyle='--')
        ax1.axvline(drop_arr.mean(), color='yellow', linewidth=2,
                    label=f'Avg: {drop_arr.mean():+.1f}%')
        pct_pos = (drop_arr > 0).mean() * 100
        ax1.set_title(f'{coin} – 30d return after a -8% day\n'
                      f'{len(drop_arr)} events  |  {pct_pos:.0f}% were profitable',
                      fontsize=11, fontweight='bold', color=color)
        ax1.set_xlabel('Return 30 days later (%)', fontsize=9)
        ax1.set_ylabel('Frequency', fontsize=9)
        ax1.legend(fontsize=9)

    # Row 2: After RSI < 30 , what happens in next 30 days
    ax2 = axes[2, ci]
    ax2.set_facecolor('#0d1117')
    rsi_arr = np.array(res['rsi_signal'])
    if len(rsi_arr) > 0:
        ax2.hist(rsi_arr, bins=25, color=color, alpha=0.8, edgecolor='#0d1117')
        ax2.axvline(0, color='white', linewidth=1.5, linestyle='--')
        ax2.axvline(rsi_arr.mean(), color='yellow', linewidth=2,
                    label=f'Avg: {rsi_arr.mean():+.1f}%')
        pct_pos_rsi = (rsi_arr > 0).mean() * 100
        ax2.set_title(f'{coin} – 30d return after RSI < 30 (oversold)\n'
                      f'{len(rsi_arr)} events  |  {pct_pos_rsi:.0f}% were profitable',
                      fontsize=11, fontweight='bold', color=color)
        ax2.set_xlabel('Return 30 days later (%)', fontsize=9)
        ax2.set_ylabel('Frequency', fontsize=9)
        ax2.legend(fontsize=9)

plt.tight_layout()
plt.savefig('08_buy_signals.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()

print("\nBuy signal summary:")
for coin in COINS:
    res      = signal_summary[coin]
    drop_arr = np.array(res['drop_signal'])
    rsi_arr  = np.array(res['rsi_signal'])
    print(f"\n  {coin}:")
    print(f"    Best month to buy  : {MONTH_NAMES[res['best_month']-1]}")
    print(f"    Worst month to buy : {MONTH_NAMES[res['worst_month']-1]}")
    print(f"    Best weeks (hist.) : {res['best_weeks']}")
    if len(drop_arr) > 0:
        print(f"    After -8% drop, avg 30d return: {drop_arr.mean():+.1f}% "
              f"({(drop_arr>0).mean()*100:.0f}% positive)")
    if len(rsi_arr) > 0:
        print(f"    After RSI<30,    avg 30d return: {rsi_arr.mean():+.1f}% "
              f"({(rsi_arr>0).mean()*100:.0f}% positive)")
print("Figure 8: Buy Signals saved.")


print("\nRunning Monte Carlo simulations 2027-2030...")

N_SIMULATIONS  = 1000
LAST_DATE      = pd.Timestamp('2026-03-27')
TARGET_YEARS   = [2027, 2028, 2029, 2030]

mc_results = {}

fig, axes = plt.subplots(1, 3, figsize=(22, 8))
fig.patch.set_facecolor('#0d1117')
fig.suptitle('Monte Carlo Price Forecast 2027–2030  (1,000 simulations per coin)',
             fontsize=15, fontweight='bold', color='white')

for ci, coin in enumerate(COINS):
    df    = data[coin].dropna(subset=['log_return'])
    color = COLORS[coin]
    ax    = axes[ci]
    ax.set_facecolor('#0d1117')

    mu     = df['log_return'].mean()
    sigma  = df['log_return'].std()
    S0     = df['close'].iloc[-1]
    N_DAYS = (pd.Timestamp('2030-12-31') - LAST_DATE).days

    all_paths = np.zeros((N_SIMULATIONS, N_DAYS))
    for sim in range(N_SIMULATIONS):
        shocks             = np.random.normal(mu, sigma, N_DAYS)
        all_paths[sim, :] = S0 * np.exp(np.cumsum(shocks))

    # Percentile bands
    p5  = np.percentile(all_paths,  5, axis=0)
    p25 = np.percentile(all_paths, 25, axis=0)
    p50 = np.percentile(all_paths, 50, axis=0)
    p75 = np.percentile(all_paths, 75, axis=0)
    p95 = np.percentile(all_paths, 95, axis=0)

    # Sample paths
    for sim in range(0, N_SIMULATIONS, 10):
        ax.plot(range(N_DAYS), all_paths[sim], color=color, alpha=0.04, linewidth=0.5)

    ax.fill_between(range(N_DAYS), p5,  p95,  alpha=0.12, color=color)
    ax.fill_between(range(N_DAYS), p25, p75,  alpha=0.22, color=color)
    ax.plot(range(N_DAYS), p50, color='white', linewidth=2,
            label=f'Median ${p50[-1]:,.2f}')

    # Year markers
    year_days = {}
    for yr in TARGET_YEARS:
        d = (pd.Timestamp(f'{yr}-12-31') - LAST_DATE).days
        if d < N_DAYS:
            ax.axvline(d, color='gray', linewidth=0.8, linestyle=':')
            ax.text(d + 5, ax.get_ylim()[1] if ax.get_ylim()[1] > 0 else p95.max(),
                    str(yr), fontsize=8, color='gray', va='top')
            idx = min(d, N_DAYS - 1)
            year_days[yr] = {
                'bear': p5[idx], 'base': p50[idx], 'bull': p95[idx]
            }

    mc_results[coin] = {'year_days': year_days, 'S0': S0,
                        'all_paths': all_paths, 'p5': p5, 'p50': p50, 'p95': p95}

    ax.set_title(f'{coin}  –  Current: ${S0:,.4f}',
                 fontsize=12, fontweight='bold', color=color)
    ax.set_xlabel('Days from March 2026', fontsize=10)
    ax.set_ylabel('Price (USD)', fontsize=10)
    ax.legend(fontsize=9)

    ann = '\n'.join([
        f"{yr}: bear ${year_days[yr]['bear']:,.2f} | "
        f"base ${year_days[yr]['base']:,.2f} | "
        f"bull ${year_days[yr]['bull']:,.2f}"
        for yr in TARGET_YEARS if yr in year_days
    ])
    ax.text(0.01, 0.99, ann, transform=ax.transAxes, fontsize=7.5,
            color='white', va='top', fontfamily='monospace',
            bbox=dict(boxstyle='round', facecolor='#1f2937', alpha=0.85))

plt.tight_layout()
plt.savefig('09_monte_carlo_2030.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print("Figure 9: Monte Carlo saved.")


fig, ax = plt.subplots(figsize=(18, 6))
fig.patch.set_facecolor('#0d1117')
ax.set_facecolor('#0d1117')
ax.axis('off')
ax.set_title('Price Forecast Table 2027–2030  –  Bear / Base / Bull Scenarios',
             fontsize=14, fontweight='bold', color='white', pad=20)

rows = []
col_labels = ['Coin', 'Current'] + [
    f'{yr} {sc}' for yr in TARGET_YEARS for sc in ['Bear','Base','Bull']
]

for coin in COINS:
    res  = mc_results[coin]
    S0   = res['S0']
    yds  = res['year_days']
    row  = [coin, f"${S0:,.4f}"]
    for yr in TARGET_YEARS:
        if yr in yds:
            row += [f"${yds[yr]['bear']:,.2f}",
                    f"${yds[yr]['base']:,.2f}",
                    f"${yds[yr]['bull']:,.2f}"]
        else:
            row += ['N/A', 'N/A', 'N/A']
    rows.append(row)

table = ax.table(cellText=rows, colLabels=col_labels,
                 loc='center', cellLoc='center')
table.auto_set_font_size(False)
table.set_fontsize(8.5)
table.scale(1.1, 2.8)

for (row, col), cell in table.get_celld().items():
    if row == 0:
        cell.set_facecolor('#1f2937')
        cell.set_text_props(color='white', fontweight='bold')
    else:
        coin_name = rows[row - 1][0]
        is_bear   = (col - 2) % 3 == 0 and col > 1
        is_bull   = (col - 2) % 3 == 2 and col > 1
        is_base   = (col - 2) % 3 == 1 and col > 1
        bg = '#1a1a2e' if is_bear else ('#1a2e1a' if is_bull else '#0d1117')
        cell.set_facecolor(bg)
        cell.set_text_props(
            color=COLORS.get(coin_name, 'white') if col == 0 else 'white'
        )
    cell.set_edgecolor('#30363d')

plt.tight_layout()
plt.savefig('10_forecast_table.png', dpi=150, bbox_inches='tight', facecolor='#0d1117')
plt.show()
print("Figure 10: Forecast Table saved.")


print("\n" + "="*70)
print("COMPLETE ANALYSIS SUMMARY")

for coin in COINS:
    df   = data[coin]
    res  = mc_results[coin]
    sig  = signal_summary[coin]
    S0   = res['S0']
    yds  = res['year_days']

    drop_arr = np.array(sig['drop_signal'])
    rsi_arr  = np.array(sig['rsi_signal'])

    print(f"\n  {coin}  (current: ${S0:,.4f})")
    print(f"  Active since       : {df['date'].min().date()}")
    print(f"  All-time high      : ${df['close'].max():,.2f}")
    print(f"  Daily volatility   : {df['return_pct'].std():.2f}%")
    print(f"  BTC correlation    : {corr_matrix.loc['BTC', coin]:.4f}")
    print(f"  Best month to buy  : {MONTH_NAMES[sig['best_month']-1]}")
    print(f"  Worst month to buy : {MONTH_NAMES[sig['worst_month']-1]}")
    if len(drop_arr) > 0:
        print(f"  Buy after -8% drop : avg +{drop_arr.mean():+.1f}% in 30d "
              f"({(drop_arr>0).mean()*100:.0f}% positive)")
    if len(rsi_arr) > 0:
        print(f"  Buy at RSI < 30    : avg +{rsi_arr.mean():+.1f}% in 30d "
              f"({(rsi_arr>0).mean()*100:.0f}% positive)")
    print(f"  --- Forecast ---")
    for yr in TARGET_YEARS:
        if yr in yds:
            print(f"    {yr}: bear ${yds[yr]['bear']:>12,.2f} | "
                  f"base ${yds[yr]['base']:>12,.2f} | "
                  f"bull ${yds[yr]['bull']:>12,.2f}")

print(f"\n  Correlation BTC-SOL : {corr_matrix.loc['BTC','SOL']:.4f}")
print(f"  Correlation BTC-TRX : {corr_matrix.loc['BTC','TRX']:.4f}")
print(f"  Correlation SOL-TRX : {corr_matrix.loc['SOL','TRX']:.4f}")
print(f"\n  TRX on BTC crash days (< -3%):")
print(f"    TRX positive: {pct_trx_pos:.1f}%   SOL positive: {pct_sol_pos:.1f}%")
```
